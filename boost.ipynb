```python
x_train = pd.read_parquet("data/train_sales_data.parquet", engine='fastparquet')
print("Training data shape:", x_train.columns)
```

```python
y_train = x_train['sales']
x_train = x_train.drop('sales', axis=1) 
print("Target variable shape:", y_train.head)
```

```python
# 1. Définition de la grille
param_grid = {
    'n_estimators': [100, 500],
    'learning_rate': [0.01, 0.1],
    'max_depth': [3, 6],
    'subsample': [0.8]
}

# 2. Initialisation
xgb = XGBRegressor(objective='reg:squarederror', n_jobs=-1)

# 3. Recherche (GridSearch ou RandomSearch)
grid_search = GridSearchCV(estimator=xgb, param_grid=param_grid, cv=3, scoring='neg_mean_squared_log_error')
grid_search.fit(x_train, y_train)

print(f"Meilleurs paramètres : {grid_search.best_params_}")
```

```python

# ============================================
# FEATURE ENGINEERING & PREPROCESSING
# ============================================

# 1. Handle datetime columns - extract useful features
if 'date' in x_train.columns:
    x_train['year'] = x_train['date'].dt.year
    x_train['month'] = x_train['date'].dt.month
    x_train['day'] = x_train['date'].dt.day
    x_train['dayofweek'] = x_train['date'].dt.dayofweek
    x_train['quarter'] = x_train['date'].dt.quarter
    x_train['weekofyear'] = x_train['date'].dt.isocalendar().week
    x_train = x_train.drop('date', axis=1)

# 2. Handle categorical columns - convert to numeric
categorical_cols = x_train.select_dtypes(include=['category', 'object']).columns

for col in categorical_cols:
    # Option A: Label Encoding (simple, works well with tree-based models)
    x_train[col] = x_train[col].astype('category').cat.codes
    
    # Option B: If you prefer one-hot encoding (uncomment below)
    # x_train = pd.get_dummies(x_train, columns=[col], prefix=col)

# 3. Ensure all columns are numeric
print("Data types after preprocessing:")
print(x_train.dtypes)
print(f"\nShape: {x_train.shape}")

# ============================================
# MODEL TRAINING
# ============================================

# Define your model
xgb = XGBRegressor(random_state=42, tree_method='hist')

# Define parameter grid
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# GridSearch
grid_search = GridSearchCV(
    estimator=xgb, 
    param_grid=param_grid, 
    cv=3, 
    scoring='neg_mean_squared_log_error',
    n_jobs=-1,  # Use all CPU cores
    verbose=2
)

# Fit the model
grid_search.fit(x_train, y_train)

print(f"\nBest parameters: {grid_search.best_params_}")
print(f"Best score: {grid_search.best_score_}")
```

```python
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.ensemble import HistGradientBoostingRegressor

# On utilise les meilleurs paramètres trouvés ou des valeurs robustes
models = {
    "XGBoost": XGBRegressor(n_estimators=500, max_depth=7, learning_rate=0.05, tree_method='hist'),
    "LightGBM": LGBMRegressor(n_estimators=500, learning_rate=0.05, num_leaves=31),
    "CatBoost": CatBoostRegressor(n_estimators=500, learning_rate=0.05, depth=7, verbose=0),
    "HistGB": HistGradientBoostingRegressor(max_iter=500, learning_rate=0.05)
}

# Boucle de test rapide
for name, model in models.items():
    model.fit(x_train, y_train)
    
    print(f"✅ {name} entraîné.")
```
