```python
import pandas as pd
import numpy as np
import os 
os.chdir("../")
os.getcwd()

# 1. Charger le fichier de résultats généré précédemment
df_results = pd.read_csv("synthese_ventes_arma.csv")

# 2. Calculer le RMSLE total (Moyenne simple)
# C'est l'indicateur standard pour évaluer la précision moyenne de tes modèles
total_rmsle = df_results['rmse_score'].mean()

print(f"📊 Résultat Global ARMA :")
print(f"Nombre de modèles évalués : {len(df_results)}")
print(f"RMSLE Moyen Total : {total_rmsle:.4f}")

# 3. Analyse par famille (pour voir où l'ARMA est le meilleur)
print("\n🔥 Top 5 des familles les mieux prédites (RMSLE min) :")
print(df_results.groupby('family')['rmse_score'].mean().sort_values().head(5))

print("\n🧊 Top 5 des familles les plus difficiles (RMSLE max) :")
print(df_results.groupby('family')['rmse_score'].mean().sort_values(ascending=False).head(5))
```
