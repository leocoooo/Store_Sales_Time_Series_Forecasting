# Objectif du projet 

L'objectif de ce projet est de prédire avec précision les ventes unitaires quotidiennes pour des milliers d'articles vendus dans les magasins de la chaîne équatorienne Corporación Favorita. Il s'agit d'un problème de prévision de séries temporelles multivariées où l'on doit anticiper la demande future en fonction de l'historique des ventes et de facteurs externes.

Le défi majeur réside dans l'intégration de variables complexes telles que les promotions, les prix du pétrole (influençant l'économie locale) et les nombreux jours fériés nationaux et régionaux.

En utilisant la métrique RMSLE, le modèle doit apprendre à capturer à la fois les tendances à long terme et les cycles de consommation hebdomadaires. La finalité est d'aider l'entreprise à optimiser sa gestion des stocks et à minimiser les pertes ou les ruptures de produits.

# 1 - Présentation du dataset 

Le cœur du projet repose sur les fichiers fournis par Kaggle, qui couvrent la période de 2013 à 2017. Le fichier principal regroupe les volumes de ventes quotidiens pour 54 magasins et 33 familles de produits, formant ainsi la cible de notre prédiction. On y trouve également l'indicateur des promotions actives pour chaque catégorie.

À cette base s'ajoutent des informations sur les points de vente, comme leur localisation géographique et leur appartenance à un groupe spécifique. Les données brutes incluent aussi le nombre de transactions par magasin et le prix quotidien du pétrole, qui sert de baromètre économique pour l'Équateur. Enfin, un calendrier liste les jours fériés nationaux et locaux ainsi que les événements spéciaux ayant marqué cette période.

Pour améliorer la précision des prévisions, plusieurs variables calculées ont été générées à partir des données sources. La date a été décomposée en plusieurs unités temporelles pour capturer les cycles de consommation selon le jour de la semaine, le mois ou le jour du mois. Ces informations permettent au modèle de repérer les pics récurrents, comme l'augmentation des ventes durant les week-ends ou lors des fins de mois.

Des  lags ont été créés pour intégrer les ventes passées comme variables explicatives du futur. Des moyennes mobiles ont également été calculées sur différentes fenêtres de temps pour lisser les variations brutales et mettre en évidence les tendances de fond. Enfin, le calendrier des jours fériés a été nettoyé pour ne conserver que les dates ayant un impact réel sur la consommation, en excluant notamment les jours fériés transférés.
